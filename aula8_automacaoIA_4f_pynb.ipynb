missão 01: Automação de Arquivos

```python
import os
import shutil

def organizar_arquivos(caminho_diretorio):
    # Garante que o caminho existe
    if not os.path.exists(caminho_diretorio):
        print(f"O caminho {caminho_diretorio} não existe.")
        return

    # Itera sobre os arquivos na pasta
    for arquivo in os.listdir(caminho_diretorio):
        caminho_completo = os.path.join(caminho_diretorio, arquivo)

        # Processa apenas arquivos (ignora pastas já existentes)
        if os.path.isfile(caminho_completo):
            nome, extensao = os.path.splitext(arquivo)

            # Define o nome da pasta baseado na extensão (ex: pdf, jpg)
            subpasta = extensao.lower().replace('.', '')

            if not subpasta:
                subpasta = 'sem_extensao'

            caminho_subpasta = os.path.join(caminho_diretorio, subpasta)

            # Cria a subpasta se ela não existir
            os.makedirs(caminho_subpasta, exist_ok=True)

            # Move o arquivo para a nova pasta
            shutil.move(caminho_completo, os.path.join(caminho_subpasta, arquivo))
            print(f"Movido: {arquivo} -> {subpasta}/")

# Exemplo de uso (ajuste o caminho conforme necessário)
caminho = './pasta_baguncada'
organizar_arquivos(caminho)
print("Organização concluída!")
```

missão 2: Consulta Simples a APIs

```python
import requests

def consultar_cep(cep):
    # URL corrigida (removido o 'x' de viacepx)
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        print(f"Consultando o CEP: {cep}...")
        response = requests.get(url, timeout=5)

        # Levanta um erro se o status code não for 200
        response.raise_for_status()

        dados = response.json()

        if "erro" in dados:
            print("CEP não encontrado.")
        else:
            print("\n--- Endereço Encontrado ---")
            print(f"Logradouro: {dados.get('logradouro')}")
            print(f"Bairro:     {dados.get('bairro')}")
            print(f"Cidade/UF:  {dados.get('localidade')}/{dados.get('uf')}")
            print(f"CEP:        {dados.get('cep')}")
            print("---------------------------")

    except requests.exceptions.RequestException as e:
        print(f"\n[ERRO DE REQUISIÇÃO]: Ocorreu um problema ao acessar a API.")
        print(f"Detalhes: {e}")

# Testando o script com o CEP fixo
meu_cep = "70293040"
consultar_cep(meu_cep)
```

Missão 03: Sistema de Notificações

```python
import time

def log_status(leitura):
    """Registra o status normal no console."""
    print(f"[LOG] Status OK: {leitura} unidades.")

def formatar_mensagem_alerta(valor):
    """Gera a string formatada para o alerta crítico."""
    return (
        "\n--- [SISTEMA DE ALERTA] ---\n"
        f"ALERTA CRÍTICO: Limite excedido!\n"
        f"Valor: {valor}\n"
        "Enviando notificação...\n"
        "---------------------------\n"
    )

def enviar_alerta_critico(valor_atual):
    """Simula o disparo do alerta."""
    mensagem = formatar_mensagem_alerta(valor_atual)
    print(mensagem)

def monitorar_sistema(limite, dados_sensor):
    """Lógica principal de monitoramento refatorada."""
    print(f"Iniciando monitoramento (Limite: {limite})...\n")

    for leitura in dados_sensor:
        if leitura > limite:
            enviar_alerta_critico(leitura)
        else:
            log_status(leitura)

        time.sleep(0.1)

# Execução
LIMITE_CRITICO = 30
leituras = [22, 25, 29, 32, 28, 35, 27]
monitorar_sistema(LIMITE_CRITICO, leituras)
```
