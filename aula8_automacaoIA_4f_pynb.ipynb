```python
import os
import shutil

# Definindo o caminho da pasta que sera organizada
# Use '.' para a pasta atual ou especifique o caminho completo
caminho_pasta = "./arquivos_baguncados"

# Criando uma pasta de exemplo para demonstracao caso nao exista
if not os.path.exists(caminho_pasta):
    os.makedirs(caminho_pasta)
    # Criando alguns arquivos vazios para testar
    open(os.path.join(caminho_pasta, "documento.pdf"), "w").close()
    open(os.path.join(caminho_pasta, "foto.jpg"), "w").close()
    open(os.path.join(caminho_pasta, "planilha.xlsx"), "w").close()

# Listando todos os arquivos na pasta
lista_arquivos = os.listdir(caminho_pasta)

# Percorrendo cada arquivo da lista
for arquivo in lista_arquivos:
    # Separando o nome da extensao
    nome, extensao = os.path.splitext(arquivo)

    # Removendo o ponto da extensao (ex: .pdf vira pdf)
    extensao = extensao.replace(".", "")

    # Verificando se e um arquivo (e nao uma pasta)
    caminho_completo_arquivo = os.path.join(caminho_pasta, arquivo)

    if os.path.isfile(caminho_completo_arquivo) and extensao != "":
        # Definindo o caminho da nova pasta baseada na extensao
        pasta_destino = os.path.join(caminho_pasta, extensao)

        # Criando a pasta de destino se ela nao existir
        if not os.path.exists(pasta_destino):
            os.makedirs(pasta_destino)

        # Movendo o arquivo para a pasta correspondente
        shutil.move(caminho_completo_arquivo, os.path.join(pasta_destino, arquivo))
        print("O arquivo " + arquivo + " foi movido para a pasta " + extensao)

print("Organizacao concluida com sucesso!")
```

### Sugestões de Funções para Manipulação de Diretórios

Para trabalhar com arquivos e pastas em Python, estas são as funções mais recomendadas:

1.  **`os.listdir(path)`**: Lista todos os arquivos e subpastas em um diretório.
2.  **`os.path.splitext(file)`**: Separa o nome do arquivo da sua extensão.
3.  **`os.path.exists(path)`**: Verifica se um arquivo ou pasta já existe.
4.  **`os.makedirs(path, exist_ok=True)`**: Cria pastas (incluindo subpastas) de forma segura.
5.  **`shutil.move(src, dst)`**: Move arquivos ou pastas de um lugar para outro.
6.  **`os.path.join(path, *paths)`**: Combina partes de caminhos de forma inteligente (ajusta as barras `/` ou `\` automaticamente).

```python
import os
import shutil

def organizar_diretorio(caminho_alvo):
    # Garante que o caminho existe
    if not os.path.exists(caminho_alvo):
        print("Caminho não encontrado!")
        return

    # Itera sobre os itens no diretório
    for item in os.listdir(caminho_alvo):
        caminho_completo = os.path.join(caminho_alvo, item)

        # Processa apenas arquivos
        if os.path.isfile(caminho_completo):
            nome, extensao = os.path.splitext(item)

            # Normaliza a extensão (remove o ponto e coloca em minúsculo)
            pasta_nome = extensao.replace(".", "").lower()

            if pasta_nome == "": # Arquivos sem extensão
                pasta_nome = "outros"

            # Define e cria a pasta de destino
            caminho_destino_pasta = os.path.join(caminho_alvo, pasta_nome)
            os.makedirs(caminho_destino_pasta, exist_ok=True)

            # Move o arquivo
            shutil.move(caminho_completo, os.path.join(caminho_destino_pasta, item))
            print("Movido: " + item + " -> " + pasta_nome)

# Executando a organização na pasta de exemplo
organizar_diretorio("./arquivos_baguncados")
print("Processo finalizado!")
```
