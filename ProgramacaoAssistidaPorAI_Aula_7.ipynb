Passo 2 - Número Primo

```python
import math

def eh_primo(n):
    if n < 2: return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
```

O Miniprojeto - Jogo da Forca

```python
import random

def validar_chute(chute, letras_tentadas):
    """Valida se a entrada é uma única letra não tentada anteriormente."""
    chute = chute.upper().strip()
    if not chute or len(chute) != 1 or not chute.isalpha():
        print('Por favor, digite apenas uma letra válida.')
        return None
    if chute in letras_tentadas:
        print(f'Você já tentou a letra {chute}!')
        return None
    return chute

def verificar_letra(chute, palavra_secreta, letras_acertadas):
    """Verifica se a letra existe na palavra e atualiza o estado do jogo."""
    acertou = False
    if chute in palavra_secreta:
        for i, letra in enumerate(palavra_secreta):
            if letra == chute:
                letras_acertadas[i] = chute
        acertou = True
    return acertou

def jogar_forca():
    palavras = ['python', 'programacao', 'computador', 'algoritmo', 'desenvolvedor']
    palavra_secreta = random.choice(palavras).upper()
    letras_acertadas = ['_' for _ in palavra_secreta]
    tentativas = 6
    letras_tentadas = []

    print('--- Jogo da Forca (Versão Refatorada) ---')

    while tentativas > 0 and '_' in letras_acertadas:
        print(f'\nPalavra: {" ".join(letras_acertadas)}')
        print(f'Tentativas: {tentativas} | Letras: {", ".join(letras_tentadas)}')

        entrada = input('Digite uma letra: ')
        chute = validar_chute(entrada, letras_tentadas)

        if chute is None:
            continue

        letras_tentadas.append(chute)

        if verificar_letra(chute, palavra_secreta, letras_acertadas):
            print(f'Boa! A letra "{chute}" pertence à palavra.')
        else:
            tentativas -= 1
            print(f'Que pena, "{chute}" não existe na palavra.')

    if '_' not in letras_acertadas:
        print(f'\nParabéns! Você venceu! A palavra era: {palavra_secreta}')
    else:
        print(f'\nGame Over! Você perdeu. A palavra era: {palavra_secreta}')

if __name__ == "__main__":
    jogar_forca()
```
