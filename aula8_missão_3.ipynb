# Sistema de Monitoramento de Temperatura Crítica
Este programa simula o monitoramento de um sensor. O objetivo é disparar um alerta apenas quando a temperatura ultrapassa um limite e o estado muda de 'Normal' para 'Crítico'.

```python
import time
import random

# --- VERSÃO 1: ESTRUTURA E IMPLEMENTAÇÃO BÁSICA ---

def verificar_condicao(leitura, limite):
    """Verifica se o valor lido ultrapassa o limite."""
    return leitura > limite

def disparar_alerta(valor):
    """Simula o envio de um alerta via console e e-mail."""
    print(f"\n[ALERTA CRÍTICO] - Temperatura atingiu {valor}°C!")
    print("Simulando envio de e-mail para o administrador... 📧")

# Teste simples da Versão 1
limite_temp = 80
leitura_atual = 85

if verificar_condicao(leitura_atual, limite_temp):
    disparar_alerta(leitura_atual)
```

### Versão Final: Refatorada e Modular

Na versão abaixo, utilizamos uma classe para manter o `estado_alerta`. Isso evita o envio de múltiplos e-mails se a temperatura continuar alta; o alerta só soa novamente se a situação for normalizada e depois voltar a ficar crítica.

```python
class MonitorTermico:
    def __init__(self, limite):
        self.limite = limite
        self.alerta_ativo = False # Controle de estado para evitar repetições

    def processar_leitura(self, valor):
        """Lógica principal de decisão de alertas."""
        if valor > self.limite:
            if not self.alerta_ativo:
                self.notificar_critico(valor)
                self.alerta_ativo = True
        else:
            if self.alerta_ativo:
                self.notificar_normalizacao(valor)
                self.alerta_ativo = False

    def notificar_critico(self, valor):
        print(f"\n>>> CRÍTICO: {valor}°C - Disparando e-mail de emergência! 🚨")

    def notificar_normalizacao(self, valor):
        print(f"\n>>> INFO: {valor}°C - Sistema normalizado. ✅")

# --- Simulação de Execução ---

monitor = MonitorTermico(limite=75)
simulacao_dados = [70, 72, 80, 82, 79, 74, 73, 85] # Oscilação de temperatura

print("Iniciando Monitoramento Automático...")
for temp in simulacao_dados:
    print(f"Leitura atual: {temp}°C", end=" | ")
    monitor.processar_leitura(temp)
    time.sleep(0.5)
```

### Como o programa funciona:
1. **Verificação**: O sistema recebe um valor numérico.
2. **Lógica Condicional**: Ele compara o valor com um `limite` pré-definido.
3. **Gestão de Estado**: A variável `alerta_ativo` armazena se o sistema já avisou sobre o problema. Se a temperatura continuar acima do limite (ex: 80, 82, 79), o alerta **não** será disparado novamente, poupando recursos e evitando spam.
4. **Normalização**: Quando o valor volta para abaixo do limite, o sistema reseta o estado e avisa que está tudo bem.
